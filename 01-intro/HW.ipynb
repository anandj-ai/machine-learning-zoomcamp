```python


url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"

df = pd.read_csv(url)
```

```python
print("Q2 - Number of records:", len(df))
```

```python
print("Q3 - Number of fuel types:", df["fuel_type"].nunique())
```

```python
print("Q4 - Columns with missing values:",

      (df.isnull().sum() > 0).sum())
```

```python
asia = df[df["origin"] == "Asia"]

print("Q5 - Max fuel efficiency in Asia:", asia["fuel_efficiency_mpg"].max())
```

```python
median_before = df["horsepower"].median()

most_frequent = df["horsepower"].mode()[0]

df["horsepower"] = df["horsepower"].fillna(most_frequent)

median_after = df["horsepower"].median()

print("Q6 - Median before:", median_before)

print("Q6 - Most frequent:", most_frequent)

print("Q6 - Median after:", median_after)
```

```python
X = asia[["vehicle_weight", "model_year"]].head(7).to_numpy()

XTX = X.T @ X

XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y

print("Q7 - w:", w)

print("Q7 - Sum of w:", w.sum())
```
